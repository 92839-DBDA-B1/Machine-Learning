### In this dataset you have to predict that whether we have to give ATTORNEY to the clients or not (0=Yes, 1=No) based on below factors.
### CLMAGE - Age of Client
### CLMSEX - Sex of client (0=Male, 1= Female)
### SEATBELT - Whether client has Seatbelt or not (0=Yes, 1=No)
### CLMINSUR - Whether Client give Insurance or not (0=Yes, 1=No)
### LOSS - Loss that client faced

### Prepare model by using Naive Bayes, KNN and SVM & Also calculate Accuracy by confusion matrix, Calculate accuracy score, precision score, recall score, f1 score.

### Dataset : Claimants.csv


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['CLMSEX']=df['CLMSEX'].fillna(df['CLMSEX'].mean())
df['CLMINSUR']=df['CLMINSUR'].fillna(df['CLMINSUR'].mean())
df['SEATBELT']=df['SEATBELT'].fillna(df['SEATBELT'].mean())
df['CLMAGE']=df['CLMAGE'].fillna(df['CLMAGE'].mean())
```

```python
df.drop(['CLMAGE'],axis=1, inplace=True)
```

```python
df['ATTORNEY'].value_counts()
```

```python
X=df.drop('ATTORNEY',axis=1)
y=df['ATTORNEY']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test= train_test_split(
    X, y, train_size=0.7, random_state=42
)
```

```python
from sklearn.naive_bayes import GaussianNB

def create_naive_bayes_model():
    model=GaussianNB()
    model.fit(X_train,y_train)
    return model
    
```

```python
from sklearn.svm import SVC

def create_svm_model():
    model=SVC()
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.neighbors import KNeighborsClassifier

def create_knn_model():
    model=KNeighborsClassifier(n_neighbors=5)
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def format_result(result):
    return f"{result * 100:.2f}"

def evaluate_model(model_info):
    model=model_info['model']
    name=model_info['name']

    y_pred=model.predict(X_test)

    accuracy=format_result(accuracy_score(y_test,y_pred))
    precision=format_result(precision_score(y_test,y_pred))
    recall=format_result(recall_score(y_test,y_pred))
    f1=format_result(f1_score(y_test,y_pred))

    return accuracy, precision, recall, f1
```

```python
models=[
    {"name": "KNN","model":create_knn_model()},
    {"name":"Naive Bayes","model":create_naive_bayes_model()},
    {"name":"SVM","model":create_svm_model()} 
]
```

```python
evaluation_result=[]
for model in models:
    accuracy, precision, recall, f1=evaluate_model(model)
    evaluation_result.append({
        "model name":model['name'],
        "accuracy":accuracy,
        "precision":precision,
        "recall":recall,
        "f1 score":f1
    })
```

```python
evaluation_result_df=pd.DataFrame(evaluation_result)
evaluation_result_df
```
