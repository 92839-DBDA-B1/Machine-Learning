```python
X=df.drop('type',axis=1)
y=df['type']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test, y_train, y_test=train_test_split(
    X,y, train_size=0.7, random_state=42
)
```

```python
from sklearn.naive_bayes import GaussianNB

def create_naive_bayes_model():
    model=GaussianNB()
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.svm import SVC

def create_svm_model():
    model=SVC()
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.neighbors import KNeighborsClassifier

def create_knn_model():
    model=KNeighborsClassifier(n_neighbors=6)
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def format_result(result):
    return f"{result * 100:.2f}"

def evaluate_model(model_info):
    model=model_info['model']
    name=model_info['name']

    y_pred=model.predict(X_test)

    accuracy=format_result(accuracy_score(y_test,y_pred))
    # precision=format_result(precision_score(y_test,y_pred,zero_division=0))
    # recall=format_result(recall_score(y_test,y_pred,zero_division=0))
    # f1=format_result(f1_score(y_test,y_pred,zero_division=0))
    precision=format_result(precision_score(y_test,y_pred,average="micro"))
    recall=format_result(recall_score(y_test,y_pred,average="micro"))
    f1=format_result(f1_score(y_test,y_pred,average="micro"))
    
    return accuracy, precision, recall, f1
```

```python
models=[
    {"name": "KNN","model":create_knn_model()},
    {"name":"Naive Bayes","model":create_naive_bayes_model()},
    {"name":"SVM","model":create_svm_model()} 
]
```

```python
evaluation_result=[]
for model in models:
    accuracy, precision, recall, f1=evaluate_model(model)
    evaluation_result.append({
        "model name":model['name'],
        "accuracy":accuracy,
        "precision":precision,
        "recall":recall,
        "f1 score":f1
    })
```

```python
evaluation_result_df=pd.DataFrame(evaluation_result)
evaluation_result_df
```
