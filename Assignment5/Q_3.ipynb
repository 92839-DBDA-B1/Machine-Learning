```python
X=df.drop('Type',axis=1)
y=df['Type']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test= train_test_split(
    X, y, train_size=0.7, random_state=42
)
```

```python
from sklearn.naive_bayes import GaussianNB

def create_naive_bayes_model():
    model=GaussianNB()
    model.fit(X_train,y_train)
    return model
    
```

```python
from sklearn.svm import SVC

def create_svm_model():
    model=SVC()
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.neighbors import KNeighborsClassifier

def create_knn_model():
    model=KNeighborsClassifier(n_neighbors=5)
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def format_result(result):
    return f"{result * 100:.2f}"

def evaluate_model(model_info):
    model=model_info['model']
    name=model_info['name']

    y_pred=model.predict(X_test)

    accuracy=format_result(accuracy_score(y_test,y_pred))
    precision=format_result(precision_score(y_test,y_pred,average="weighted",zero_division=0))
    recall=format_result(recall_score(y_test,y_pred,average="weighted",zero_division=0))
    f1=format_result(f1_score(y_test,y_pred,average="weighted",zero_division=0))

    return accuracy, precision, recall, f1
```
