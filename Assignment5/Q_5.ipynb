###### In Churn.csv dataset you have to predict that customer is exited or not based on various factors given in dataset.
###### Implement a KNN,SVM and Naive Bayes model and calculate Accuracy by confusion matrix, Calculate accuracy score, precision score, recall score, f1 score.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Churn.csv')
df.head(2)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.drop(['RowNumber','CustomerId','Surname','HasCrCard'],axis=1, inplace=True)
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder=LabelEncoder()
df['Geography']=encoder.fit_transform(df['Geography'])
df['Gender']=encoder.fit_transform(df['Gender'])
```

```python
df.head(2)
```

```python
df.corr()
```

```python
df['Exited'].value_counts()
```

```python
X=df.drop('Exited',axis=1)
y=df['Exited']
```

```python
from imblearn.over_sampling import SMOTE
smote=SMOTE(random_state=42)
X_new, y_new=smote.fit_resample(X,y)
```

```python
#y_new.value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test, y_train, y_test=train_test_split(
    X_new,y_new, train_size=0.7, random_state=42
)
```

```python
from sklearn.naive_bayes import GaussianNB

def create_naive_bayes_model():
    model=GaussianNB()
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.svm import SVC

def create_svm_model():
    model=SVC()
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.neighbors import KNeighborsClassifier

def create_knn_model():
    model=KNeighborsClassifier(n_neighbors=6)
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def format_result(result):
    return f"{result * 100:.2f}"

def evaluate_model(model_info):
    model=model_info['model']
    name=model_info['name']

    y_pred=model.predict(X_test)

    accuracy=format_result(accuracy_score(y_test,y_pred))
    precision=format_result(precision_score(y_test,y_pred))
    recall=format_result(recall_score(y_test,y_pred))
    f1=format_result(f1_score(y_test,y_pred))
    
    return accuracy, precision, recall, f1
```

```python
models=[
    {"name": "KNN","model":create_knn_model()},
    {"name":"Naive Bayes","model":create_naive_bayes_model()},
    {"name":"SVM","model":create_svm_model()} 
]
```

```python
evaluation_result=[]
for model in models:
    accuracy, precision, recall, f1=evaluate_model(model)
    evaluation_result.append({
        "model name":model['name'],
        "accuracy":accuracy,
        "precision":precision,
        "recall":recall,
        "f1 score":f1
    })
```

```python
evaluation_result_df=pd.DataFrame(evaluation_result)
evaluation_result_df
```
