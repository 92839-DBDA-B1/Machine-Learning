```python
from sklearn.linear_model import LogisticRegressionCV
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

def model_random_forest():
    model = RandomForestClassifier(n_estimators=50)
    model.fit(X_train,y_train)
    return model
    
def model_decision_tree():
    model = DecisionTreeClassifier(max_depth=10)
    model.fit(X_train,y_train)
    return model

def model_xgb_classifier():
    model = XGBClassifier()
    model.fit(X_train,y_train)
    return model

def model_cat_boost():
    model = CatBoostClassifier(iterations=20)
    model.fit(X_train,y_train)
    return model
    
def model_logistic_regression():
    model = LogisticRegressionCV(max_iter=2000)
    model.fit(X_train,y_train)
    return model
def model_naive_bayes():
    model = GaussianNB()
    model.fit(X_train,y_train)
    return model
def model_knn():
    model = KNeighborsClassifier()
    model.fit(X_train,y_train)
    return model
def model_svm():
    model = SVC()
    model.fit(X_train,y_train)
    return model
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,classification_report,confusion_matrix

def format_metric(metric):
    return f"{metric*100:2f}"

def evaluation_metrics(models_info):
    name=models_info['name']
    model=models_info['model']

    y_pred = model.predict(X_test)
    # accuracy=f"{accuracy_score(y_test,y_pred) * 100:.2f}"
    
    accuracy=format_metric(accuracy_score(y_test,y_pred))
    # precision=format_metric(precision_score(y_test,y_pred,average="weighted"))
    # recall=format_metric(recall_score(y_test,y_pred,average="weighted"))
    # f1=format_metric(f1_score(y_test,y_pred,average="weighted"))
    cm=confusion_matrix(y_test,y_pred)

    precision = format_metric(precision_score(y_test, y_pred, average="weighted", zero_division=0))
    recall = format_metric(recall_score(y_test, y_pred, average="weighted", zero_division=0))
    f1 = format_metric(f1_score(y_test, y_pred, average="weighted", zero_division=0))

  

    return accuracy,precision,recall,f1,cm
    
```

```python
evaluation_result=[]

for model in models:
    accuracy,precision,recall,f1,cm = evaluation_metrics(model)
    
    accuracy_using_cm = f"{(cm.trace() / cm.sum()) * 100:.2f}"

    evaluation_result.append(
        {'Model_Name':model['name'],
         'Accuracy':accuracy,
         'Precision':precision,
         'Recall':recall,
         'f1':f1,
         'Accuracy using cm':accuracy_using_cm
        }
    )
 
    
evaluation_results = pd.DataFrame(evaluation_result)
evaluation_results
```
