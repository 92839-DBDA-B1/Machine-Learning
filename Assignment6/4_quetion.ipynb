```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
```

```python
X=df.drop('type',axis=1)
y=encoder.fit_transform(df['type'])
```

```python
from imblearn.over_sampling import SMOTE
# smote = SMOTE(random_state=42)  -  
## error ValueError: Expected n_neighbors <= n_samples_fit, but n_neighbors = 5, n_samples_fit = 4, n_samples = 4

smote = SMOTE(random_state=42,k_neighbors=1)



from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,train_size=0.7,random_state=42)
X_train,y_train=smote.fit_resample(X_train,y_train)
```

```python
from sklearn.linear_model import LogisticRegressionCV
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

def model_random_forest():
    model = RandomForestClassifier(n_estimators=50)
    model.fit(X_train,y_train)
    return model
    
def model_decision_tree():
    model = DecisionTreeClassifier(max_depth=10)
    model.fit(X_train,y_train)
    return model

def model_xgb_classifier():
    model = XGBClassifier()
    model.fit(X_train,y_train)
    return model

def model_cat_boost():
    model = CatBoostClassifier(iterations=20)
    model.fit(X_train,y_train)
    return model
    
def model_logistic_regression():
    model = LogisticRegressionCV(max_iter=2000)
    model.fit(X_train,y_train)
    return model
def model_naive_bayes():
    model = GaussianNB()
    model.fit(X_train,y_train)
    return model
def model_knn():
    model = KNeighborsClassifier()
    model.fit(X_train,y_train)
    return model
def model_svm():
    model = SVC()
    model.fit(X_train,y_train)
    return model
```

```python
models = [
   
    {'name':'Random Forest','model':model_random_forest()},
    {'name':'Decision Tree','model':model_decision_tree()},
    {'name':'XGB','model':model_xgb_classifier()},
    {'name':'CatBoost','model':model_cat_boost()}
]
```

```python
def format_metric(metric):
    return f"{metric *100:.2f}"
```

```python
from sklearn.metrics import accuracy_score,precision_score,f1_score,recall_score,classification_report,confusion_matrix
def evaluation_metric(model_info):
    name=model_info['name']
    model=model_info['model']
    y_pred=model.predict(X_test)
    accuracy = format_metric(accuracy_score(y_test,y_pred))
    precision = format_metric(precision_score(y_test,y_pred,average="weighted",zero_division=0))
    recall= format_metric(recall_score(y_test,y_pred,average="weighted",zero_division=0))
    f1= format_metric(f1_score(y_test,y_pred,average="weighted",zero_division=0))
    cm=confusion_matrix(y_test,y_pred)
    return accuracy,precision,recall,f1,cm

evaluation_result=[]

for model in models:
    accuracy,precision,recall,f1,cm = evaluation_metric(model)
    
    accuracy_using_cm = f"{(cm.trace() / cm.sum()) * 100:.2f}"

    evaluation_result.append(
        {'Model_Name':model['name'],
         'Accuracy':accuracy,
         'Precision':precision,
         'Recall':recall,
         'f1':f1,
         # 'Confusion_Matrix':cm,
         'Accuracy_using_cm':accuracy_using_cm
    
        }
    )
 
    
evaluation_results = pd.DataFrame(evaluation_result)
evaluation_results

# Convert to DataFrame
evaluation_results = pd.DataFrame(evaluation_result)
evaluation_results
    
    
    
```
