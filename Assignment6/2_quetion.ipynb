```python

missing_cols =['cigsPerDay','BPMeds','totChol','BMI','heartRate','glucose']

for i in missing_cols:
    df[i]=df[i].fillna(df[i].mean())
```

```python
df.drop(['education'],axis=1,inplace=True)
```

```python
X=df.drop('TenYearCHD',axis=1)
y=df['TenYearCHD']
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_new,y_new = smote.fit_resample(X,y)
```

#### splitting the dataset

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X_new,y_new,train_size=0.7,random_state=42)
```

#### building the model

```python
from sklearn.linear_model import LogisticRegressionCV
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

def model_random_forest():
    model = RandomForestClassifier(n_estimators=100)
    model.fit(X_train,y_train)
    return model
    
def model_decision_tree():
    model = DecisionTreeClassifier(max_depth=10)
    model.fit(X_train,y_train)
    return model
    
def model_xgb_classifier():
    model = XGBClassifier(n_estimators=50)
    model.fit(X_train,y_train)
    return model

def model_cat_boost():
    model = CatBoostClassifier(n_estimators=50)
    model.fit(X_train,y_train)
    return model
    
def model_logistic_regression():
    model = LogisticRegressionCV(max_iter=2000)
    model.fit(X_train,y_train)
    return model
def model_naive_bayes():
    model = GaussianNB()
    model.fit(X_train,y_train)
    return model
def model_knn():
    model = KNeighborsClassifier()
    model.fit(X_train,y_train)
    return model
def model_svm():
    model = SVC()
    model.fit(X_train,y_train)
    return model
```

```python
models = [
    {'name':'Logistic Regression','model':model_logistic_regression()},
    {'name':'Naivye Bayes', 'model':model_naive_bayes()},
    {'name':'KNN','model':model_knn()},
    {'name':'SVM','model':model_svm()},
    {'name':'Random Forest','model':model_random_forest()},
    {'name':'Decision Tree','model':model_decision_tree()},
    {'name':'XGB','model':model_xgb_classifier()},
    {'name':'CatBoost','model':model_cat_boost()}
]


```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,classification_report,confusion_matrix

def format_metric(metric):
    return f"{metric*100:2f}"

def evaluation_metrics(models_info):
    name=models_info['name']
    model=models_info['model']

    y_pred = model.predict(X_test)
    accuracy=f"{accuracy_score(y_test,y_pred) * 100:.2f}"
    
    accuracy=format_metric(accuracy_score(y_test,y_pred))
    precision=format_metric(precision_score(y_test,y_pred))
    recall=format_metric(recall_score(y_test,y_pred))
    f1=format_metric(f1_score(y_test,y_pred))
    cm=confusion_matrix(y_test,y_pred)
  

    return accuracy,precision,recall,f1,cm
    
```

```python
evaluation_result=[]

for model in models:
    accuracy,precision,recall,f1,cm = evaluation_metrics(model)

    # tn, fp, fn, tp = cm.ravel()
    # accuracy_using_cm= f"{((tp + tn) / (tp + tn + fp + fn))*100:.2f}"    
    accuracy_using_cm = f"{(cm.trace() / cm.sum()) * 100:.2f}"

    evaluation_result.append(
        {'Model_Name':model['name'],
         'Accuracy':accuracy,
         'Precision':precision,
         'Recall':recall,
         'f1':f1,
         'Accuracy_using_cm':accuracy_using_cm
        }
    )
 
    
evaluation_results = pd.DataFrame(evaluation_result)
evaluation_results

```
