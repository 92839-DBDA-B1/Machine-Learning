Implement a KNN,SVM and Naive Bayes model and calculate Accuracy by confusion matrix, Calculate accuracy score, precision score, recall score, f1 score.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Datasets/Churn.csv")
```

```python
df.drop(['CustomerId','Surname'],inplace=True,axis=1)
```

```python
object_col = df.select_dtypes(include="object").columns
from sklearn.preprocessing import LabelEncoder
for col in object_col:
    encoder = LabelEncoder()
    df[col]=encoder.fit_transform(df[col])
df.info()
```

```python
X=df.drop('Exited',axis=1)
y=df['Exited']
```

```python
from imblearn.over_sampling import SMOTE
# smote = SMOTE(random_state=42)  -  
## error ValueError: Expected n_neighbors <= n_samples_fit, but n_neighbors = 5, n_samples_fit = 4, n_samples = 4

# smote = SMOTE(random_state=42,k_neighbors=3)
smote = SMOTE(random_state=42,k_neighbors=3)
X_new,y_new=smote.fit_resample(X,y)

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_new,y_new,train_size=0.7,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegressionCV
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

def model_random_forest():
    model = RandomForestClassifier(n_estimators=50)
    model.fit(X_train,y_train)
    return model
    
def model_decision_tree():
    model = DecisionTreeClassifier(max_depth=10)
    model.fit(X_train,y_train)
    return model

def model_xgb_classifier():
    model = XGBClassifier()
    model.fit(X_train,y_train)
    return model

def model_cat_boost():
    model = CatBoostClassifier(iterations=20)
    model.fit(X_train,y_train)
    return model
    
def model_logistic_regression():
    model = LogisticRegressionCV(max_iter=2000)
    model.fit(X_train,y_train)
    return model
def model_naive_bayes():
    model = GaussianNB()
    model.fit(X_train,y_train)
    return model
def model_knn():
    model = KNeighborsClassifier()
    model.fit(X_train,y_train)
    return model
def model_svm():
    model = SVC()
    model.fit(X_train,y_train)
    return model

```

```python
models = [
   
    {'name':'Random Forest','model':model_random_forest()},
    {'name':'Decision Tree','model':model_decision_tree()},
    {'name':'XGB','model':model_xgb_classifier()},
    {'name':'CatBoost','model':model_cat_boost()}
]
```

```python
def format_metric(metric):
    return f"{metric *100:.2f}"
```

```python
from sklearn.metrics import accuracy_score,precision_score,f1_score,recall_score,classification_report,confusion_matrix
def evaluation_metric(model_info):
    name=model_info['name']
    model=model_info['model']
    y_pred=model.predict(X_test)
    accuracy = format_metric(accuracy_score(y_test,y_pred))
    precision = format_metric(precision_score(y_test,y_pred))
    recall= format_metric(recall_score(y_test,y_pred))
    f1= format_metric(f1_score(y_test,y_pred))
    cm=confusion_matrix(y_test,y_pred)
    return accuracy,precision,recall,f1,cm
```

```python
evaluation_result=[]

for model in models:
    accuracy, precision, recall, f1, cm = evaluation_metric(model)
    tn, fp, fn, tp = cm.ravel()
    accuracy_using_cm= f"{((tp + tn) / (tp + tn + fp + fn))*100:.2f}"

    evaluation_result.append({
    'Model_Name': model['name'],
    'Accuracy': accuracy,
    'Precision': precision,
    'Recall': recall,
    'F1-Score': f1,
    'Accuracy_using_cm': accuracy_using_cm
})


# Convert to DataFrame
evaluation_results = pd.DataFrame(evaluation_result)
evaluation_results
    
    
    
```
