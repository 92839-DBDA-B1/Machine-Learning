### slit the data

```python
X=df.drop('GPA',axis=1)

y=df['GPA']
```

from sklearn.model_selection import train_test_split
#train size deremine the train set
#randam_state it is used to split the same data ererytime
X_train,X_test,y_train,y_test=train_test_split(
    X,y,train_size=0.7, test_size=0.3,random_state=42)
X_train

### data pre-processing

```python
####-science the data is already is grete shape, noneed to shape
```

### train the model

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
X_train.shape
```

```python
X_train.
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
mae=mean_absolute_error(y_test,y_pred)
print(f"mean absolute error={mae:2f}")
```

```python
from sklearn.metrics import  mean_absolute_percentage_error
mape=mean_absolute_percentage_error(y_test,y_pred)
print(f"mean_absolute_percentage_error={mape}")
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred)
print(f"mean squared error={mse}")
```

```python
from sklearn.metrics import root_mean_squared_error
rmse=root_mean_squared_error(y_test,y_pred)
print(f"root mean squared error={rmse}")
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test,y_pred)
print(f"r2 score={r2:.2f}")
```

### inferene the model

```python
# predict GPA of a student having SAT as 10000
result=model.predict([[1000],[1500]])
result
```

```python
# in general, the model is represented as
# y=B0+B1X1+E
# the B0 is a constant
B0=model.intercept_
print(B0)
```

```python
B1=model.coef_[0]
B1
```

```python
SAT_SCORE=1000
y=B0+B1*SAT_SCORE
print(y)
```

```python
SAT_SCORE=1500
y=B0+B1*SAT_SCORE
print(y)
```
