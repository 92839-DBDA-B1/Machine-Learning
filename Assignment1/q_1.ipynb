The delivery_time.csv dataset contains information about Delivery time and Sorting Time. We have to prepare linear regression model to predict Sorting time on the basis of Delivery Time. Also evaluate the model by using evaluation metrics. Do proper EDA & visualize the result.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("delivery_time.csv")
df.head()
```

```python
df.info() 
```

```python
df.describe() 
```

```python
df.head()
```

```python
df.corr()  
```

```python
sns.heatmap(df.corr(),annot=True)
```

```python
plt.scatter(df['Delivery Time'],df['Sorting Time'],color="red",label="Sorting")
plt.title("Delivery VS Sortng")
plt.xlabel("Delivery Time")
plt.ylabel("Sorting Time")
plt.legend()
```

```python
X=df.drop('Sorting Time',axis=1)
y=df['Sorting Time']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    X, y,train_size=0.7,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
mae=mean_absolute_error(y_test,y_pred)
print(f"mean absolut error={mae:.2f}")
```

```python
from sklearn.metrics import mean_absolute_percentage_error
mape=mean_absolute_percentage_error(y_test,y_pred)
print(f"mean aboslute percentage error={mape:.2f}")
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred)
print(f"mean squared error={mse:.2f}")
```

```python
from sklearn.metrics import mean_squared_log_error
msle=mean_squared_log_error(y_test,y_pred)
print(f"mean squared log error={msle:.2f}")
```

```python
from sklearn.metrics import root_mean_squared_log_error
rmsle=root_mean_squared_log_error(y_test,y_pred)
print(f"root mean squared log error={rmsle:.2f}")
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test,y_pred)
print(f"R-squared value={r2:.2f}")
```

```python
plt.scatter(X_test['Delivery Time'],y_test,label="sorting Time")

plt.plot(X_test['Delivery Time'],y_pred,color="red",label="egression Line")
plt.scatter(X_test['Delivery Time'],y_pred,color="red",marker="s")
plt.title("Regression Result")
plt.xlabel("Delivery in Time")
plt.ylabel("Sorting in time")
plt.legend()
plt.grid(visible=True,alpha=0.3)
```
