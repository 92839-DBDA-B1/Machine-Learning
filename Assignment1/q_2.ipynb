The Newspaperdata.csv dataset contains information of nwespapers with their daily sales and sunday sales. We have to predict sunday sales of newspaper on the basis of daily sales of newspaper. For that build linear regression model and evaluate the model by using evaluation metrics. Do proper EDA & visualize the result.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_1=pd.read_csv("NewspaperData.csv")
df=df_1.drop('Newspaper',axis=1)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(),annot=True)
```

```python
plt.scatter(df['daily'],df['sunday'],color="red",label="Newspaper")
plt.title("dailt newspapers")
plt.xlabel("daily")
plt.ylabel("sunday newspaper")
plt.legend()
```

```python
X=df.drop("sunday",axis=1)
y=df["sunday"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(
    X, y, train_size=0.7,random_state=42)
```

### train the model

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error

mae=mean_absolute_error(y_test,y_pred)
print(f"mean absolute error={mae:.2f}")
```

```python
from sklearn.metrics import mean_absolute_percentage_error
mape=mean_absolute_percentage_error(y_test,y_pred)
print(f"mean absolute percentage error={mape:.2f}")
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred)
print(f"mean squared error={mse:.2f}")
```

```python
from sklearn.metrics import  mean_squared_log_error
msle=mean_squared_log_error(y_test,y_pred)
print(f"mean squared log error={msle:.2f}")
```

```python
from sklearn.metrics import root_mean_squared_error

rmse=root_mean_squared_error(y_test,y_pred)
print(f"root mean squared error={rmse:.2f}")
```

```python
from sklearn.metrics import root_mean_squared_log_error
rmsle=root_mean_squared_log_error(y_test,y_pred)
print(f"root mean squared log error={rmsle:.2f}")
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test,y_pred)
print(f"r2 score={r2:.2f}")
```

```python
plt.scatter(X_test['daily'],y_test,label="salary")


plt.plot(X_test['daily'],y_pred,color="blue",label="Sunday Newspaper")
plt.scatter(X_test['daily'],y_pred,color="blue",marker="s")

plt.title("Newspaper prediction")
plt.xlabel("daily")
plt.ylabel("sunday")
plt.legend()
plt.grid(visible=True,alpha=0.3)
```
