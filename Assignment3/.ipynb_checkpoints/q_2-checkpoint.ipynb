```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("SAT-GPA.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(), annot=True)
```

```python
plt.scatter(df['SAT'],df['GPA'],color="red",label="SAT VS GPA")
plt.title("SAT AND GPA")
plt.xlabel("SAT")
plt.ylabel("GPA")
plt.legend()
```

```python
X=df.drop('GPA', axis=1)
y=df['GPA']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    X,y,train_size=0.7, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)
print(f"mean absolute error = {mae:.2f}")
```

```python
from sklearn.metrics import mean_absolute_percentage_error

# calculate mean absolute percentage error
mape = mean_absolute_percentage_error(y_test, y_pred)
print(f"mean absolute percentage error = {mape:.2f}")
```

```python
from sklearn.metrics import mean_squared_error

# calculage mean squared error 
mse = mean_squared_error(y_test, y_pred)
print(f"mean squared error = {mse:.2f}")
```

```python
from sklearn.metrics import mean_squared_log_error

# calculate mean squared log error
msle = mean_squared_log_error(y_test, y_pred)
print(f"mean squared log error = {msle:.2f}")
```

```python
from sklearn.metrics import root_mean_squared_error

# calculate root mean squared error
rmse = root_mean_squared_error(y_test, y_pred)
print(f"root mean squared error: {rmse:.2f}")
```

```python
from sklearn.metrics import root_mean_squared_log_error

# calculate root mean squared log error
rmsle = root_mean_squared_log_error(y_test, y_pred)
print(f"root mean squared log error = {rmsle:.2f}")
```

```python
from sklearn.metrics import r2_score

# calculate R-squared value
r2 = r2_score(y_test, y_pred)
print(f"R-squared value = {r2:.2f}")
```

```python
plt.scatter(X_test['SAT'],y_test,label="SAT")
plt.plot(X_test['SAT'], y_pred, color="red", label="Regression Line")
plt.scatter(X_test['SAT'], y_pred, color="red", marker="s")

plt.title("Regression Result")
plt.xlabel("SAT")
plt.ylabel("GPA")
plt.legend()
plt.grid(visible=True, alpha=0.3)
```
